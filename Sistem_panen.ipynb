```python
def hitung_total_hasil(panen):
    return sum(panen)


hasil_panen = [100, 150, 200]

total = hitung_total_hasil(hasil_panen)

print("=== Sistem Pencatatan Hasil Panen ===")
print("Data hasil panen:", hasil_panen, "kg")
print("Total hasil panen:", total, "kg")
```
